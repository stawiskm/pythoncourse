## Simplest Perceptron Learning Algorithm

Implement the Perceptron Learning Algorithm to classify data points into two classes (0 or 1). The algorithm should update the weights based on the misclassified points until all points are correctly classified. The perceptron should take two input features and output a binary classification.

### Simple Perceptron and how it works:
- The perceptron is a simple algorithm that can be used to classify data points into two classes.
  
- It takes input features (x1, x2) and weights (w1, w2) to calculate the weighted sum.
    - Meaning: `weighted_sum = x1*w1 + x2*w2`
  
- The weighted sum is then passed through an activation function (e.g., step function) to produce the output (0 or 1).
    - Meaning: `output = 1 if weighted_sum >= threshold else 0`
    - The threshold is a parameter that determines the decision boundary.
    - If the weighted sum is greater than or equal to the threshold, the output is 1; otherwise, it is 0.
    - The threshold can be thought of as a measure of how confident the perceptron is in its classification.
    - A higher threshold means the perceptron is more confident in its classification.
    - A lower threshold means the perceptron is less confident in its classification.
  
- The weights are updated based on the misclassified points to improve the classification accuracy
    - Meaning: `w_new = w_old + learning_rate * (expected_output - output) * input`
    - The learning rate is a parameter that controls how much the weights are updated.
    - A higher learning rate means the weights are updated more aggressively.
    - A lower learning rate means the weights are updated more conservatively.
    - The learning rate is a hyperparameter that needs to be tuned to achieve the best performance.
    - The process of updating the weights based on the misclassified points is known as the perceptron learning algorithm.
   
- The process is repeated.
    - Meaning: `repeat until  number of iterations is reached`
  
- The perceptron learning algorithm is a simple way to learn a linear decision boundary between two classes.
- It can be used for binary classification tasks where the data is linearly separable.
- The algorithm is based on the concept of supervised learning, where the correct class labels are provided for training.
- The perceptron learning algorithm is an example of a single-layer neural network with a linear activation function.
- It is a building block for more complex neural network architectures.
- The perceptron learning algorithm is a simple and intuitive way to learn a linear decision boundary between two classes.
