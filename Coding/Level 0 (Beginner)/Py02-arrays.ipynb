## 6. Excurse: Images

### Images as Arrays

In many scientific and engineering fields, images are represented as arrays of pixel values.

**Example**: Load and display an image using numpy and matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np
import imageio

# TODO: Load an image into a numpy array
image = imageio.imread('path/to/image.jpg')  # TODO: Specify the correct path to the image
plt.imshow(image)
plt.show()
```

### Image Processing with Numpy

You can perform various image processing tasks using numpy arrays, such as changing brightness, contrast, etc.

**Example**: Convert an image to grayscale.

```python
# TODO: Convert the loaded image to grayscale
gray_image = np.dot(image[...,:3], [0.2989, 0.5870, 0.1140])  # Convert to grayscale
plt.imshow(gray_image, cmap='gray')
plt.show()
```

### Displaying Images

You can use matplotlib to display images stored in numpy arrays.

**Example**: Display an image with a title.

```python
# TODO: Display the grayscale image with the title 'Grayscale Image'
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Image')  # TODO: Add the correct title
plt.show()
```
