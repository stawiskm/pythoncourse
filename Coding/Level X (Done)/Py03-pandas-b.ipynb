## Data Structures in Pandas
### Creating a Series
A Pandas Series is a one-dimensional array-like object. Complete the TODO to create a Series.

````python
# TODO: Create a Pandas Series
data = [10, 20, 30, 40]
series = pd.Series(data)
print(series)
````

### Creating a DataFrame
A Pandas DataFrame is a two-dimensional data structure. Complete the TODO to create a DataFrame.

````python
# TODO: Create a Pandas DataFrame
data = {
    'Column1': [1, 2, 3, 4],
    'Column2': [5, 6, 7, 8]
}
df = pd.DataFrame(data)
print(df)
````

## Basic Operations
### Viewing Data
You can view the first few rows of a DataFrame using the `head` method. Complete the TODO to view the data.

````python
# TODO: View the first few rows of the DataFrame
print(df.head())
````

### Selecting Data
Select a specific column from the DataFrame. Complete the TODO to select the 'Column1'.

````python
# TODO: Select 'Column1' from the DataFrame
column1 = df['Column1']
print(column1)
````

### Filtering Data
Filter rows based on a condition. Complete the TODO to filter rows where 'Column1' is greater than 2.

````python
# TODO: Filter rows where 'Column1' > 2
filtered_df = df[df['Column1'] > 2]
print(filtered_df)
````

### Adding and Modifying Columns
Add a new column to the DataFrame. Complete the TODO to add a 'Column3' with values.

````python
# TODO: Add 'Column3' to the DataFrame
df['Column3'] = [9, 10, 11, 12]
print(df)
````

### Deleting Columns
Delete a column from the DataFrame. Complete the TODO to drop 'Column2'.

````python
# TODO: Drop 'Column2' from the DataFrame
df = df.drop('Column2', axis=1)
print(df)
````

### Handling Missing Data
Handle missing data by filling or dropping. Complete the TODO to fill missing values with 0.

````python
# TODO: Fill missing values with 0
df = df.fillna(0)
print(df)
````

## Advanced Operations
### Grouping Data
Group data by a column and calculate aggregate statistics. Complete the TODO to group by 'Column1' and calculate the mean.

````python
# TODO: Group by 'Column1' and calculate the mean
grouped_df = df.groupby('Column1').mean()
print(grouped_df)
````

### Merging DataFrames
Merge two DataFrames on a common column. Complete the TODO to merge `df` and `df2` on 'Column1'.

````python
# Create another DataFrame
data2 = {
    'Column1': [1, 2, 3, 4],
    'Column4': ['A', 'B', 'C', 'D']
}
df2 = pd.DataFrame(data2)

# TODO: Merge df and df2 on 'Column1'
merged_df = pd.merge(df, df2, on='Column1')
print(merged_df)
````

## Saving and Loading Data
### Reading Data from a File
Read a CSV file into a DataFrame. Complete the TODO to read a file named 'data.csv'.

````python
# TODO: Read data from 'data.csv'
df_from_csv = pd.read_csv('data.csv')
print(df_from_csv)
````

### Writing Data to a File
Write a DataFrame to a CSV file. Complete the TODO to write `df` to 'output.csv'.

````python
# TODO: Write the DataFrame to 'output.csv'
df.to_csv('output.csv', index=False)
````
