## BMI Calculator

Create a program that calculates a person's Body Mass Index (BMI) based on their height and weight. The program should categorize the BMI according to standard weight status categories (e.g., underweight, normal weight, overweight, obesity).

### TODOs:
1. Write a function to calculate BMI.
2. Categorize the BMI based on standard weight status categories.
3. Test the function with different inputs.

```python
# Function to calculate BMI
def calculate_bmi(weight, height):
    # TODO: Calculate BMI using the formula: BMI = weight (kg) / (height (m))^2
    bmi = weight / (height ** 2)
    return bmi

# Function to categorize BMI
def categorize_bmi(bmi):
    # TODO: Categorize BMI based on the standard weight status categories, by using the following values:
    # Underweight: BMI < 18.5
    # Normal weight: 18.5 <= BMI < 25
    # Overweight: 25 <= BMI < 30
    # Obesity: BMI >= 30
    category = None
    if bmi < 18.5:
        category = "Underweight"
    elif 18.5 <= bmi < 25:
        category = "Normal weight"
    elif 25 <= bmi < 30:
        category = "Overweight"
    else:
        category = "Obesity"
    return category

# Test the BMI calculator
weight = 70  # kg
height = 1.75  # meters

bmi = calculate_bmi(weight, height)
category = categorize_bmi(bmi)

print(f"BMI: {bmi:.2f}")
print(f"Category: {category}")
```

## Temperature Converter

Write a program that converts temperature between Celsius, Fahrenheit, and Kelvin.

### TODOs:

```python
# Function to convert temperature from Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    # TODO: Convert Celsius to Fahrenheit
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

# Function to convert temperature from Fahrenheit to Celsius
def fahrenheit_to_celsius(fahrenheit):
    # TODO: Convert Fahrenheit to Celsius
    celsius = (fahrenheit - 32) * 5/9
    return celsius

# Function to convert temperature from Celsius to Kelvin
def celsius_to_kelvin(celsius):
    #TODO: Convert Celsius to Kelvin
    kelvin = celsius + 273.15
    return kelvin

# Function to convert temperature from Kelvin to Celsius
def kelvin_to_celsius(kelvin):
    #TODO: Convert Kelvin to Celsius
    celsius = kelvin - 273.15
    return celsius

# Function to convert temperature from Fahrenheit to Kelvin
def fahrenheit_to_kelvin(fahrenheit):
    #TODO: Convert Fahrenheit to Kelvin
    kelvin = (fahrenheit - 32) * 5/9 + 273.15
    return kelvin

# Function to convert temperature from Kelvin to Fahrenheit
def kelvin_to_fahrenheit(kelvin):
    #TODO: Convert Kelvin to Fahrenheit
    fahrenheit = (kelvin - 273.15) * 9/5 + 32
    return fahrenheit

# Test the temperature conversion functions
celsius = 37.5
fahrenheit = celsius_to_fahrenheit(celsius)
print(f"{celsius}°C = {fahrenheit:.2f}°F")

fahrenheit = 100.0
celsius = fahrenheit_to_celsius(fahrenheit)
print(f"{fahrenheit}°F = {celsius:.2f}°C")

celsius = 100.0
kelvin = celsius_to_kelvin(celsius)
print(f"{celsius}°C = {kelvin:.2f}K")

kelvin = 373.15
celsius = kelvin_to_celsius(kelvin)
print(f"{kelvin}K = {celsius:.2f}°C")

fahrenheit = 100.0
kelvin = fahrenheit_to_kelvin(fahrenheit)
print(f"{fahrenheit}°F = {kelvin:.2f}K")

kelvin = 310.928
fahrenheit = kelvin_to_fahrenheit(kelvin)
print(f"{kelvin}K = {fahrenheit:.2f}°F")
```

```python
# Function which converts the given temperature from one unit ("C","F","K") to another
def convert_temperature(temperature, from_unit, to_unit):
    # First Convert the temperature to Celsius 
    if from_unit == "C":
        celsius = temperature
    elif from_unit == "F":
        celsius = fahrenheit_to_celsius(temperature)
    elif from_unit == "K":
        celsius = kelvin_to_celsius(temperature)
    else:
        raise ValueError(f"Unknown unit {from_unit}")

    # Then convert Celsius to the desired unit
    if to_unit == "C":
        converted_temperature = celsius
    elif to_unit == "F":
        converted_temperature = celsius_to_fahrenheit(celsius)
    elif to_unit == "K":
        converted_temperature = celsius_to_kelvin(celsius)
    else:
        raise ValueError(f"Unknown unit {to_unit}")

    return converted_temperature

# Test the temperature conversion function
temperature = 100.0
from_unit = "F"
to_unit = "C"
converted_temperature = convert_temperature(temperature, from_unit, to_unit)
print(f"{temperature}{from_unit} = {converted_temperature}{to_unit}")
```

2. Test the function with different sets of medications.

```python
# Function for drug interaction checker
def check_interaction(medication1, medication2):
    # List of known drug interactions
    interactions = {
        ("aspirin", "ibuprofen"): "Avoid concomitant use",
        ("paracetamol", "ibuprofen"): "Safe to use together",
        ("paracetamol", "aspirin"): "Safe to use together"
    }

    # Check if the combination is in the interactions dictionary
    interaction = interactions.get((medication1, medication2))
    if interaction:
        return interaction
    else:
        return "No known interaction"
```

## Basic DNA Sequence Analyzer

Write a program that analyzes a DNA sequence provided by the user. The program should count the occurrences of each nucleotide (A, T, C, G) and identify simple patterns or motifs within the sequence.

### TODOs:
1. Write a functions to analyze a DNA sequence.
   1. count_nucleotides(sequence)
   2. find_motifs(sequence)
2. Test the function with different DNA sequences.

```python
# Function to analyze DNA sequence
def analyze_dna_count(dna_sequence):
    # Count the number of nucleotides
    counts = {
        "A": 0,
        "C": 0,
        "G": 0,
        "T": 0
    }

    for nucleotide in dna_sequence:
        # Increment the count for the nucleotide
        if nucleotide in counts:
            counts[nucleotide] += 1
        else:
            # Ignore unknown nucleotides
            pass

    return counts
```

```python
# Function to analyze DNA sequence find motifs
def analyze_dna_motifs(dna_sequence, motif):
    # Find the positions of the motif in the DNA sequence
    positions = []
    i = 0
    while i < len(dna_sequence):
        # Find the next occurrence of the motif using find method: find(motif, start) - returns -1 if not found 
        position = dna_sequence.find(motif, i)
        if position != -1:
            # Add the position to the list
            positions.append(position)
            # Move to the next position
            i = position + 1
        else:
            # No more occurrences found
            break

    return positions
```

```python
# Test the DNA analysis functions
dna_sequence = "ACGTACGTACGT"
counts = analyze_dna_count(dna_sequence)
print(counts)

motif = "CG"
positions = analyze_dna_motifs(dna_sequence, motif)
print(positions)
```

## Patient Record System

Create a simple patient record system that allows users to add, update, and view patient information. The system should store information such as patient name and age, and provide options to add/update additional details.

### TODOs:
1. Create a class to represent a patient record.
   1. Implement methods to add, update, and view patient information.
2. Test the class by creating patient records and updating/viewing information.

```python
# Class to represent a patient record
class PatientRecord:
    def __init__(self, name, age, gender, medical_history):
        # TODO: Initialize the patient record with the provided information
        self.name = name
        self.age = age
        self.gender = gender
        self.medical_history = medical_history
    
    def __str__(self):
        # Return a string representation of the patient record
        return f"Name: {self.name}, Age: {self.age}, Gender: {self.gender}, Medical History: {self.medical_history}"
    
    def __repr__(self):
        # Return a string representation of the patient record for debugging
        return f"PatientRecord({self.name}, {self.age}"
    

# List to store patient records
patient_records = []

# Function to add a patient record
def add_patient(name, age, gender, medical_history):
    #TODO: Create a new patient record and add it to the list
    patient = PatientRecord(name, age,gender, medical_history)
    patient_records.append(patient)

# Function to update a patient record
def update_patient(name, age=None, gender=None, medical_history=None):
    # Find the patient by name and update the record
    for patient in patient_records:
        if patient.name == name:
            # Update the patient record with the provided information
            if age is not None:
                patient.age = age
            if gender is not None:
                patient.gender = gender
            if medical_history is not None:
                patient.medical_history = medical_history
    

# Function to delete a patient record
def delete_patient(name):
    # Find the patient by name and delete the record
    for patient in patient_records:
        if patient.name == name:
            # Remove the patient record from the list
            patient_records.remove(patient)
    

# Function to search for a patient by name
def search_patient(name):
    # Find the patient by name and return the record
    for patient in patient_records:
        if patient.name == name:
            # Return the patient record
            return patient
    

# Test the Patient Record Management System
add_patient('John Doe', 30, 'Male', 'None')
add_patient('Jane Smith', 25, 'Female', 'Asthma')

print('All Patients:', patient_records)  # Update to print in a more readable format

update_patient('John Doe', medical_history='Diabetes')
print('Updated John Doe:', search_patient('John Doe'))  # Update to print in a more readable format

delete_patient('Jane Smith')
print('After Deleting Jane Smith:', patient_records)  # Update to print in a more readable format

search_result = search_patient('John Doe')
print('Search for John Doe:', search_result)  # Update to print in a more readable format
```

## Simplest Perceptron Learning Algorithm

Implement the Perceptron Learning Algorithm to classify data points into two classes (0 or 1). The algorithm should update the weights based on the misclassified points until all points are correctly classified. The perceptron should take two input features and output a binary classification.

### Simple Perceptron and how it works:
- The perceptron is a simple algorithm that can be used to classify data points into two classes.
  
- It takes input features (x1, x2) and weights (w1, w2) to calculate the weighted sum.
    - Meaning: `weighted_sum = x1*w1 + x2*w2`
  
- The weighted sum is then passed through an activation function (e.g., step function) to produce the output (0 or 1).
    - Meaning: `output = 1 if weighted_sum >= threshold else 0`
    - The threshold is a parameter that determines the decision boundary.
    - If the weighted sum is greater than or equal to the threshold, the output is 1; otherwise, it is 0.
    - The threshold can be thought of as a measure of how confident the perceptron is in its classification.
    - A higher threshold means the perceptron is more confident in its classification. 
    - A lower threshold means the perceptron is less confident in its classification.
  
- The weights are updated based on the misclassified points to improve the classification accuracy
    - Meaning: `w_new = w_old + learning_rate * (expected_output - output) * input`
    - The learning rate is a parameter that controls how much the weights are updated.
    - A higher learning rate means the weights are updated more aggressively.
    - A lower learning rate means the weights are updated more conservatively.
    - The learning rate is a hyperparameter that needs to be tuned to achieve the best performance.
    - The process of updating the weights based on the misclassified points is known as the perceptron learning algorithm.
   
- The process is repeated.
    - Meaning: `repeat until  number of iterations is reached`
  
- The perceptron learning algorithm is a simple way to learn a linear decision boundary between two classes.
- It can be used for binary classification tasks where the data is linearly separable.
- The algorithm is based on the concept of supervised learning, where the correct class labels are provided for training.
- The perceptron learning algorithm is an example of a single-layer neural network with a linear activation function.
- It is a building block for more complex neural network architectures.
- The perceptron learning algorithm is a simple and intuitive way to learn a linear decision boundary between two classes.

### TODOs:
1. Implement the Perceptron Learning Algorithm.
   1. Initialize weights and bias.
   2. Update weights based on misclassified points.
   3. Repeat until all points are correctly classified.
2. Test the algorithm with sample data points.

```python
# Implement the Perceptron Learning Algorithm to classify data points into two classes (0 or 1). 
# The perceptron should take two input features and output a binary classification.

# Perceptron class
class Perceptron:
    def __init__(self, learning_rate=0.1, num_epochs=100):
        self.learning_rate = learning_rate
        self.num_epochs = num_epochs
        self.weights = [0, 0]
        self.bias = 0

    def predict(self, x):
        # Calculate the weighted sum of inputs
        weighted_sum = self.weights[0] * x[0] + self.weights[1] * x[1] + self.bias
        # Apply the step function
        if weighted_sum >= 0:
            return 1
        else:
            return 0

    def train(self, X, y):
        for epoch in range(self.num_epochs):
            for i in range(len(X)):
                # Make a prediction
                prediction = self.predict(X[i])
                # Update the weights and bias
                self.weights[0] += self.learning_rate * (y[i] - prediction) * X[i][0]
                self.weights[1] += self.learning_rate * (y[i] - prediction) * X[i][1]
                self.bias += self.learning_rate * (y[i] - prediction)
```

### OR Gate Example:

```python
# Training data
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 0, 0, 1]

# Create and train the perceptron
perceptron = Perceptron()

perceptron.train(X, y)

# Test the perceptron
for i in range(len(X)):
    prediction = perceptron.predict(X[i])
    print(f"Input: {X[i]}, Prediction: {prediction}")   
```
