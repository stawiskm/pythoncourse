## Exercise 9: File Handling Error

The code below is supposed to read a file and print its contents, but it has an error. Try to fix it. If you don't have a file named `sample.txt`, you may create one or change the filename to an existing file.

```python
# Fix the file handling error below

# file = open('sample.txt', 'r')
# content = file.read()
# print(content)
# file.close()

# Fix the import error below
try:
    with open('sample.txt', 'w') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("File not found")
except PermissionError:
    print("Permission denied")
except Exception as e:
    print("An error occurred:", e)
```

## Exercise 10: Function Error

The code below defines a function to calculate the factorial of a number, but it contains an error. Try to find and fix it.

```python
# Fix the function error below

def factorial(n):
    if n == 0:
        return 1
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))
```

## Exercise 11: Exception Handling Error

This code is supposed to handle exceptions during division, but it contains an error. Can you fix it so that it handles zero division correctly?

```python
# Fix the exception handling error below

try:
    numerator = 10
    denominator = 0
    result = numerator / denominator
    print("The result is", result)
except ZeroDivisionError:
    print("Error: Division by zero!")
except Exception as e:
    print("An error occurred:", e)
```
