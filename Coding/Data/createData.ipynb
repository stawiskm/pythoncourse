# Generate CSV Files for Medical Data Analysis Problems

This notebook generates synthetic CSV datasets for the medical data analysis problems described in the previous notebook.

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random
```

## Problem 1: ICU Patient Monitoring

```python
# Generate ICU dataset
num_patients = 100
num_records_per_patient = 50
patients = []
timestamps = []
heart_rates = []
blood_pressures = []
respiratory_rates = []
oxygen_saturations = []

start_time = datetime.now()

for i in range(num_patients):
    patient_id = i + 1
    for _ in range(num_records_per_patient):
        patients.append(patient_id)
        timestamps.append(start_time + timedelta(minutes=random.randint(0, 60*24)))
        heart_rates.append(random.randint(60, 120))
        blood_pressures.append(random.randint(80, 140))
        respiratory_rates.append(random.randint(12, 20))
        oxygen_saturations.append(random.uniform(88, 100))

icu_df = pd.DataFrame({
    'patient_id': patients,
    'timestamp': timestamps,
    'heart_rate': heart_rates,
    'blood_pressure': blood_pressures,
    'respiratory_rate': respiratory_rates,
    'oxygen_saturation': oxygen_saturations
})

# Save to CSV
icu_df.to_csv('icu_data.csv', index=False)
icu_df.head()
```

## Problem 2: Analyzing Clinical Test Results

```python
# Generate Clinical Test Results dataset
tests = ['Glucose', 'Cholesterol', 'Hemoglobin', 'WBC', 'RBC']
reference_ranges = {
    'Glucose': '70-140',
    'Cholesterol': '100-200',
    'Hemoglobin': '12-16',
    'WBC': '4-11',
    'RBC': '4.5-6'
}

patients = []
test_names = []
test_results = []
reference_ranges_list = []
test_dates = []

for i in range(num_patients):
    patient_id = i + 1
    for test in tests:
        patients.append(patient_id)
        test_names.append(test)
        reference_ranges_list.append(reference_ranges[test])
        test_results.append(random.uniform(0.5 * float(reference_ranges[test].split('-')[0]),
                                           1.5 * float(reference_ranges[test].split('-')[1])))
        test_dates.append(start_time.date() + timedelta(days=random.randint(0, 30)))

clinical_tests_df = pd.DataFrame({
    'patient_id': patients,
    'test_name': test_names,
    'test_result': test_results,
    'reference_range': reference_ranges_list,
    'test_date': test_dates
})

# Save to CSV
clinical_tests_df.to_csv('clinical_tests.csv', index=False)
clinical_tests_df.head()
```

## Problem 3: Predicting Hospital Bed Occupancy Rates

```python
# Generate Hospital Bed Occupancy dataset
days = 365
total_beds = 200

dates = [start_time.date() + timedelta(days=i) for i in range(days)]
occupied_beds = [random.randint(100, total_beds) for _ in range(days)]
admissions = [random.randint(10, 30) for _ in range(days)]
discharges = [random.randint(5, 25) for _ in range(days)]

bed_df = pd.DataFrame({
    'date': dates,
    'total_beds': total_beds,
    'occupied_beds': occupied_beds,
    'admissions': admissions,
    'discharges': discharges
})

# Save to CSV
bed_df.to_csv('bed_occupancy.csv', index=False)
bed_df.head()
```

## Problem 4: Patient Satisfaction Survey Analysis

```python
# Generate Patient Satisfaction Survey dataset
departments = ['Cardiology', 'Neurology', 'Orthopedics', 'Pediatrics', 'Oncology']
genders = ['Male', 'Female']

patient_ids = []
ages = []
genders_list = []
departments_list = []
satisfaction_scores = []
wait_times = []
treatment_outcomes = []

for i in range(num_patients):
    patient_ids.append(i + 1)
    ages.append(random.randint(18, 85))
    genders_list.append(random.choice(genders))
    departments_list.append(random.choice(departments))
    satisfaction_scores.append(random.randint(1, 10))
    wait_times.append(random.randint(5, 120))
    treatment_outcomes.append(random.choice(['Improved', 'No Change', 'Worsened']))

satisfaction_df = pd.DataFrame({
    'patient_id': patient_ids,
    'age': ages,
    'gender': genders_list,
    'department': departments_list,
    'satisfaction_score': satisfaction_scores,
    'wait_time': wait_times,
    'treatment_outcome': treatment_outcomes
})

# Save to CSV
satisfaction_df.to_csv('satisfaction_survey.csv', index=False)
satisfaction_df.head()
```

## Problem 5: Tracking Medication Side Effects

```python
# Generate Medication Side Effects dataset
medications = ['MedA', 'MedB', 'MedC']
side_effects_list = ['Nausea', 'Dizziness', 'Headache', 'Fatigue', 'Insomnia', 'None']
severity_levels = ['Mild', 'Moderate', 'Severe']

patient_ids = []
ages = []
genders_list = []
medications_list = []
side_effects = []
severities = []
durations = []

for i in range(num_patients):
    patient_ids.append(i + 1)
    ages.append(random.randint(18, 85))
    genders_list.append(random.choice(genders))
    medications_list.append(random.choice(medications))
    reported_side_effects = ', '.join(random.sample(side_effects_list, random.randint(1, 3)))
    side_effects.append(reported_side_effects)
    severities.append(random.choice(severity_levels))
    durations.append(random.randint(1, 30))

side_effects_df = pd.DataFrame({
    'patient_id': patient_ids,
    'age': ages,
    'gender': genders_list,
    'medication': medications_list,
    'side_effects': side_effects,
    'severity': severities,
    'duration': durations
})

# Save to CSV
side_effects_df.to_csv('medication_side_effects.csv', index=False)
side_effects_df.head()
```

## Problem 6: Evaluating Treatment Outcomes in a Clinical Trial

```python
# Generate Clinical Trial dataset
treatments = ['Treatment A', 'Treatment B']
severity_levels = ['Mild', 'Moderate', 'Severe']
outcomes = ['Improved', 'No Change', 'Worsened']

patient_ids = []
ages = []
genders_list = []
treatments_list = []
baseline_severities = []
outcomes_list = []
follow_up_durations = []

for i in range(num_patients):
    patient_ids.append(i + 1)
    ages.append(random.randint(18, 85))
    genders_list.append(random.choice(genders))
    treatments_list.append(random.choice(treatments))
    baseline_severities.append(random.choice(severity_levels))
    outcomes_list.append(random.choice(outcomes))
    follow_up_durations.append(random.randint(7, 90))

trial_df = pd.DataFrame({
    'patient_id': patient_ids,
    'age': ages,
    'gender': genders_list,
    'treatment': treatments_list,
    'baseline_severity': baseline_severities,
    'outcome': outcomes_list,
    'follow_up_duration': follow_up_durations
})

# Save to CSV
trial_df.to_csv('clinical_trial.csv', index=False)
trial_df.head()
```
